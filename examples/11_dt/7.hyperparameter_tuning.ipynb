# Hyperparameter Tuning with GridSearchCV

## Overview

**Hyperparameter tuning** systematically searches for the best combination of hyperparameters using cross-validation. **GridSearchCV** exhaustively searches over a specified parameter grid.

## Why Hyperparameter Tuning?

- Default parameters rarely give optimal results
- Different datasets require different configurations
- Proper tuning significantly improves model performance
- Prevents both underfitting and overfitting

## Learning Objectives

1. Understand what hyperparameters are
2. Use GridSearchCV for systematic tuning
3. Tune Decision Trees, Bagging, and Boosting
4. Interpret tuning results
5. Select best parameters based on cross-validation

## Step 1: Import Libraries

```python
from matplotlib import pyplot as plt
from sklearn.datasets import load_iris, load_diabetes
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.tree import DecisionTreeClassifier, DecisionTreeRegressor
from sklearn.ensemble import BaggingClassifier, BaggingRegressor
from sklearn.ensemble import GradientBoostingClassifier, GradientBoostingRegressor
from sklearn.metrics import accuracy_score, r2_score
import pandas as pd
import numpy as np
```

## Section 1: Decision Tree Classification Tuning

```python
iris_ds = load_iris()
X_iris = iris_ds.data
y_iris = iris_ds.target

X_train_iris, X_test_iris, y_train_iris, y_test_iris = train_test_split(
    X_iris, y_iris, test_size=0.2, random_state=42
)

dt_param_grid = {
    "criterion": ["gini", "entropy"],
    "max_depth": [3, 5, 7],
    "min_samples_split": [2, 5],
    "min_samples_leaf": [1, 2]
}

dt_clf = DecisionTreeClassifier(random_state=42)
dt_grid_search = GridSearchCV(
    estimator=dt_clf,
    param_grid=dt_param_grid,
    cv=5,
    scoring="accuracy",
    n_jobs=-1
)

print("Tuning Decision Tree Classifier...")
dt_grid_search.fit(X_train_iris, y_train_iris)

print(f"Best Parameters: {dt_grid_search.best_params_}")
print(f"Best CV Score: {dt_grid_search.best_score_:.4f}")

y_pred_dt = dt_grid_search.predict(X_test_iris)
test_acc_dt = accuracy_score(y_test_iris, y_pred_dt)
print(f"Test Accuracy: {test_acc_dt:.4f}")
```

## Section 2: Decision Tree Regression Tuning

```python
diabetes_ds = load_diabetes()
X_diabetes = diabetes_ds.data
y_diabetes = diabetes_ds.target

X_train_diab, X_test_diab, y_train_diab, y_test_diab = train_test_split(
    X_diabetes, y_diabetes, test_size=0.2, random_state=42
)

dt_reg_param_grid = {
    "max_depth": [3, 5, 7],
    "min_samples_split": [2, 5],
    "min_samples_leaf": [1, 2]
}

dt_reg = DecisionTreeRegressor(random_state=42)
dt_reg_grid_search = GridSearchCV(
    estimator=dt_reg,
    param_grid=dt_reg_param_grid,
    cv=5,
    scoring="r2",
    n_jobs=-1
)

print("Tuning Decision Tree Regressor...")
dt_reg_grid_search.fit(X_train_diab, y_train_diab)

print(f"Best Parameters: {dt_reg_grid_search.best_params_}")
print(f"Best CV Score (R²): {dt_reg_grid_search.best_score_:.4f}")

y_pred_dt_reg = dt_reg_grid_search.predict(X_test_diab)
test_r2_dt_reg = r2_score(y_test_diab, y_pred_dt_reg)
print(f"Test R² Score: {test_r2_dt_reg:.4f}")
```

## Section 3: Bagging Classifier Tuning

```python
bagging_clf_param_grid = {
    "n_estimators": [5, 10, 20],
    "max_samples": [0.8, 1.0],
    "max_features": [0.8, 1.0]
}

bagging_clf = BaggingClassifier(
    estimator=DecisionTreeClassifier(max_depth=5),
    random_state=42
)
bagging_grid_search = GridSearchCV(
    estimator=bagging_clf,
    param_grid=bagging_clf_param_grid,
    cv=5,
    scoring="accuracy",
    n_jobs=-1
)

print("Tuning Bagging Classifier...")
bagging_grid_search.fit(X_train_iris, y_train_iris)

print(f"Best Parameters: {bagging_grid_search.best_params_}")
print(f"Best CV Score: {bagging_grid_search.best_score_:.4f}")

y_pred_bagging = bagging_grid_search.predict(X_test_iris)
test_acc_bagging = accuracy_score(y_test_iris, y_pred_bagging)
print(f"Test Accuracy: {test_acc_bagging:.4f}")
```

## Section 4: Bagging Regressor Tuning

```python
bagging_reg_param_grid = {
    "n_estimators": [5, 10, 20],
    "max_samples": [0.8, 1.0],
    "max_features": [0.8, 1.0]
}

bagging_reg = BaggingRegressor(
    estimator=DecisionTreeRegressor(max_depth=5),
    random_state=42
)
bagging_reg_grid_search = GridSearchCV(
    estimator=bagging_reg,
    param_grid=bagging_reg_param_grid,
    cv=5,
    scoring="r2",
    n_jobs=-1
)

print("Tuning Bagging Regressor...")
bagging_reg_grid_search.fit(X_train_diab, y_train_diab)

print(f"Best Parameters: {bagging_reg_grid_search.best_params_}")
print(f"Best CV Score (R²): {bagging_reg_grid_search.best_score_:.4f}")

y_pred_bagging_reg = bagging_reg_grid_search.predict(X_test_diab)
test_r2_bagging_reg = r2_score(y_test_diab, y_pred_bagging_reg)
print(f"Test R² Score: {test_r2_bagging_reg:.4f}")
```

## Section 5: Gradient Boosting Classifier Tuning

```python
gb_clf_param_grid = {
    "n_estimators": [50, 100],
    "learning_rate": [0.05, 0.1],
    "max_depth": [3, 5],
    "subsample": [0.8, 1.0]
}

gb_clf = GradientBoostingClassifier(random_state=42)
gb_grid_search = GridSearchCV(
    estimator=gb_clf,
    param_grid=gb_clf_param_grid,
    cv=5,
    scoring="accuracy",
    n_jobs=-1
)

print("Tuning Gradient Boosting Classifier...")
gb_grid_search.fit(X_train_iris, y_train_iris)

print(f"Best Parameters: {gb_grid_search.best_params_}")
print(f"Best CV Score: {gb_grid_search.best_score_:.4f}")

y_pred_gb = gb_grid_search.predict(X_test_iris)
test_acc_gb = accuracy_score(y_test_iris, y_pred_gb)
print(f"Test Accuracy: {test_acc_gb:.4f}")
```

## Section 6: Gradient Boosting Regressor Tuning

```python
gb_reg_param_grid = {
    "n_estimators": [50, 100],
    "learning_rate": [0.05, 0.1],
    "max_depth": [3, 5],
    "subsample": [0.8, 1.0]
}

gb_reg = GradientBoostingRegressor(random_state=42)
gb_reg_grid_search = GridSearchCV(
    estimator=gb_reg,
    param_grid=gb_reg_param_grid,
    cv=5,
    scoring="r2",
    n_jobs=-1
)

print("Tuning Gradient Boosting Regressor...")
gb_reg_grid_search.fit(X_train_diab, y_train_diab)

print(f"Best Parameters: {gb_reg_grid_search.best_params_}")
print(f"Best CV Score (R²): {gb_reg_grid_search.best_score_:.4f}")

y_pred_gb_reg = gb_reg_grid_search.predict(X_test_diab)
test_r2_gb_reg = r2_score(y_test_diab, y_pred_gb_reg)
print(f"Test R² Score: {test_r2_gb_reg:.4f}")
```

## Section 7: Model Comparison

```python
clf_results = pd.DataFrame({
    "Model": ["Decision Tree", "Bagging", "Gradient Boosting"],
    "Test Accuracy": [test_acc_dt, test_acc_bagging, test_acc_gb]
})

print("\nClassification Results (Iris):")
print(clf_results.to_string(index=False))

reg_results = pd.DataFrame({
    "Model": ["Decision Tree", "Bagging", "Gradient Boosting"],
    "Test R²": [test_r2_dt_reg, test_r2_bagging_reg, test_r2_gb_reg]
})

print("\nRegression Results (Diabetes):")
print(reg_results.to_string(index=False))
```

## Section 8: Visualize Results

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

models = clf_results["Model"]
test_accs = clf_results["Test Accuracy"]
axes[0].bar(models, test_accs, color=["blue", "green", "red"], alpha=0.7)
axes[0].set_ylabel("Test Accuracy")
axes[0].set_title("Classification: Test Accuracy")
axes[0].set_ylim([0, 1])

models = reg_results["Model"]
test_r2s = reg_results["Test R²"]
axes[1].bar(models, test_r2s, color=["blue", "green", "red"], alpha=0.7)
axes[1].set_ylabel("Test R² Score")
axes[1].set_title("Regression: Test R²")
axes[1].set_ylim([0, 1])

plt.tight_layout()
plt.show()
```

## Key Takeaways

1. **GridSearchCV systematically searches** parameter grid with cross-validation
2. **Cross-validation prevents overfitting** by evaluating on multiple splits
3. **Best parameters vary by dataset** - no universal optimal configuration
4. **Computational cost** increases exponentially with number of hyperparameters
5. **Scoring metric matters:** accuracy for classification, R² for regression
6. **Gradient Boosting** typically outperforms other methods
7. **Always validate** on held-out test set after tuning
8. **Trade-offs exist:** Better CV scores don't always mean better test performance
